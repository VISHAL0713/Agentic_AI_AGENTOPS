### 2. Configure the API key
For a temporary learning notebook, you can enter the key without saving it in the notebook. In production, keep secrets in environment variables or a secret manager.

```python
import os
from getpass import getpass

if not os.environ.get("OPENAI_API_KEY"):
    api_key = os.environ["OPENAI_API_KEY"] = getpass("OpenAI API key: ")
    print(f"OpenAI API key set in environment variable OPENAI_API_KEY: {api_key[0:4]}")
```

### 3. Create a chat model
A chat model is the part of an agent that reasons over the current conversation and produces the next AI message. temperature=0 makes the output more consistent while learning.
If your account does not have access to this model, replace the model name with one available to you.

```python
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model_name="gpt-4.1-mini", temperature=0)
model
```

### 4. Model invoke with one message

Invoking the model using invoke() function

```python
response = model.invoke("Explain LLM in one sentence.")
print(response)
print("\n")
print(response.content)
```

### 5. Stream a response
stream() yields chunks as they arrive. Streaming is useful in an agent UI because the user can see progress before the full response is ready.

```python
for chunk in model.stream("Write a three-line poem about debugging."):
    print(chunk.content, end="", flush=True)
print()
```

### 6. Ask for structured output
Free-form text is useful for conversation. For an agent that needs reliable data to act on, define a schema and use with_structured_output().

```python
from pydantic import BaseModel, Field

class LearningPlan(BaseModel):
    topic:str
    steps:list[str]
    estimated_time: int = Field(description="Estimated time in hours")
    
model.with_structured_output(LearningPlan)
response = model.invoke("Create a learning plan for learning LangChain.")
print(response)
print("\n")
print(response.content)
```
